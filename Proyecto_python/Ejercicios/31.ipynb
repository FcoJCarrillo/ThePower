```python
"""
Crea una función que solicite al usuario ingresar una lista de nombres y luego solicite un nombre para buscar en
esa lista. Si el nombre está en la lista, se imprime un mensaje indicando que fue encontrado, de lo contrario, se
lanza una excepción.
"""
def buscar_nombre():
    nombres = input("Introduce nombres separados por comas: ").split(",")

    nombre = input("Nombre a buscar: ")

    if nombre in nombres:
        print("Nombre encontrado")
    else:
        raise Exception("Nombre no encontrado")


try:
    buscar_nombre()
except Exception as e:
    print(e)
```
