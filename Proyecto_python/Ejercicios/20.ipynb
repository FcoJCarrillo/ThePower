```python
"""
Para una lista con elementos tipo integer y string obtén una nueva lista sólo con los valores int. Usa la función
filter()
"""

lista = [3,"Adios",6,"Holaa",10,49,395,623,"aaa"]
def solo_enteros(lista):
    return list(filter(lambda x: isinstance(x, int), lista))

print(solo_enteros(lista))
```
