```python
"""
Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.
"""
mascostas = ["Mapache", "Oso","Tigre", "Serpiente Pitón","Mapache", "Cocodrilo", "Oso"]
def primer_duplicado(lista):
    vistos = set()

    for elemento in lista:
        if elemento in vistos:
            return elemento
        
        vistos.add(elemento)
    return None
print(f"El primer valor duplicado de la lista {mascostas} es {primer_duplicado(mascostas)}")
```
