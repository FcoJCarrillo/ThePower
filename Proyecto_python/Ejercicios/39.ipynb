```python
"""
Escribe un programa que determine qué calificación en texto tiene un alumno en base a su calificación numérica.
Las reglas de calificación son:
- 0 - 69 insuficiente
- 70 - 79 bien
- 80 - 89 muy bien
- 90 - 100 excelente
"""
nota = int(input("Introduce la nota: "))

if 0 <= nota <= 69:
    print("Insuficiente")

elif 70 <= nota <= 79:
    print("Bien")

elif 80 <= nota <= 89:
    print("Muy bien")

elif 90 <= nota <= 100:
    print("Excelente")

else:
    print("Nota inválida")
```
