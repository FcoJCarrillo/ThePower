```python
"""
Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map()
"""
lista_1 = [1,2,3,4,5]
lista_2 = [6,7,8,9,10]

def resta_listas(num_1, num_2):
    """
    Una función

    Args:
        num_1 (int): Minuendo de la resta
        num_2 (int): Sustraendo de la resta

    Returns:
        int: devuelve el resultado de la resta
    """
    return num_1 - num_2

resultado = list(map(resta_listas,lista_1,lista_2))
print(f"El resultado de la resta de las dos listas son es: {resultado}")
```
