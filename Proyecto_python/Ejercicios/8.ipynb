```python
"""
Escribe un programa que pida al usuario dos números e intente dividirlos. Si el usuario ingresa un valor no numérico
o intenta dividir por cero, maneja esas excepciones de manera adecuada. Asegúrate de mostrar un mensaje
indicando si la división fue exitosa o no.
"""
def division(num_1, num_2):
    """
    La función lo que hace es dividir dos números y restorna el resultado la división.

    Args:
        num_1 (int): _description_
        num_2 (int): _description_

    Returns:
        STRING: Retorna el resultado de la división si ambos son numeros o muestra un mensaje que no es puede devidir si el divisor es 0 o
        introducido algún caracter que no sea un número. 
    """
    try:
        num_1 = float(num_1)
        num_2 = float(num_2)
        resultado = num_1 / num_2
        return f"La división fue exitosa. Resultado: {resultado}"
    except ValueError:
        return f"Uno de los dos valores introducidos no es un número"
    except ZeroDivisionError:
        return "No se puede dividir entre 0"

numero_1 = input("Esciba el numero del dividendo de la función: ")
numero_2 = input("Esciba el numero del divisor de la función: ")

division(numero_1,numero_2)
```
