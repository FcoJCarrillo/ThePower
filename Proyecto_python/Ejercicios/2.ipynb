```python
"""
Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map()
"""
lista_numeros = [1,2,3,4,5,6,7,8,9]
def sqr(num):
    """
    Función para sacar el doble de una lista de numeros
    Args:
        num (list): Es una lista de números

    Returns:
        int: Nos retorna el doble del número
    """
    return num*2
resultado = list(map(sqr,lista_numeros))
print(f"El resultado del doble de la lista es: {resultado}")
```
