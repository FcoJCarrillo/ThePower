```python
"""
Crea una función que retorne las palabras de una lista de palabras que comience con una letra en especifico. Usa la
función filter()
"""
def palabras_por_letra(lista_palabras, letra):
    return list(filter(lambda p: p.lower().startswith(letra.lower()), lista_palabras))

lista = ["Perro", "Gato", "Mapache", "Loro", "Oso", "Conejo"]
print(palabras_por_letra(lista, "ñ"))
```
