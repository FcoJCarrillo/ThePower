```python
"""
Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras
pero en diferente orden.
"""

def buscar_nombre():
    nombres = input("Introduce nombres separados por comas: ").split(",")

    nombre = input("Nombre a buscar: ")

    if nombre in nombres:
        print("Nombre encontrado")
    else:
        raise Exception("Nombre no encontrado")

try:
    buscar_nombre()
except Exception as e:
    print(e)
```
