```python
"""
Concatena una lista de palabras.Usa la función reduce() .
"""
from functools import reduce

mascostas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]

def concatenar(lista):
    return reduce(lambda x, y: x + y, lista)

print(f"Concatenando la lista {mascostas} en {concatenar(mascostas)}")
```
