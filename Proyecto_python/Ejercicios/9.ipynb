```python
"""
Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista
excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre",
"Serpiente Pitón", "Cocodrilo", "Oso"].Usa la función filter()
"""

def filtrar_mascotas(mascotas):
    mascostas_excluidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]
    return list(filter(lambda mascota: mascota not in mascostas_excluidas, mascotas))


mascotas = ["Perro", "Gato", "Mapache", "Loro", "Oso", "Conejo"]

resultado = filtrar_mascotas(mascotas)

print(resultado)

```
