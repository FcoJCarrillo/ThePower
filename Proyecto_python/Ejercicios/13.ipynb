```python
"""
Genera una función la cual, para un conjunto de caracteres, devuelva una lista de tuplas con cada letra en
mayúsculas y minúsculas. Las letras no pueden estar repetidas .Usa la función map()
"""
def letras(palabras):
    letras_unicas = set(c for c in palabras if c.isalpha())
    resultado = map(lambda letra: (letra.upper(), letra.lower()), letras_unicas)
    return list(resultado)

texto = "CabraijHUBoliadnuiuHBI"
print(letras(texto))
```
