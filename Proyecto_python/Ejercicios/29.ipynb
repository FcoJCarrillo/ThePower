```python
"""
Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el
carácter '#', excepto los últimos cuatro.
"""
texto = "Prueba para el ejecicio número 29, caracateres"
def enmascarar(texto):
    texto = str(texto)

    if len(texto) <= 4:
        return texto

    return "#" * (len(texto)-4) + texto[-4:]

print(f"El texto enmascarado es {enmascarar(texto)}")
```
